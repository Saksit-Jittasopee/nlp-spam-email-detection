```python
from transformers import pipeline
```

```python
classifier = pipeline("text-classification", model="./spam_detector_model", tokenizer="./spam_detector_model")
```

```python
# Message change this to test the model with different emails
new_emails = [
    "🔥Congratulations! You have been selected for a free gift.",
    "Hey, can we reschedule our 1:1 to next Tuesday?"
]
```

```python
predictions = classifier(new_emails)
print(predictions)
```
